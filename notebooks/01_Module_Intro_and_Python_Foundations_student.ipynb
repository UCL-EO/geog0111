## 2.1 Why notebooks?

A notebook combines **text**, **code** and **outputs** in one document. This makes an analysis easy to follow and to reproduce. A single notebook can, for example:

- download a climate station record and compute annual mean temperature;
- load a city or catchment boundary;
- retrieve Sentinel-2 or Landsat imagery for that area and compute NDVI or land surface temperature;
- compare two years, or fit a simple model to the time series;
- present the maps, figures and your interpretation.

That is the structure of your Part B project.

We use **Google Colab** because it runs in the browser, needs no installation and gives everyone the same environment.

## 2.2 Notebook environments and Google Colab

A notebook is made of **cells**. The two main types are:

| Cell type | Purpose |
|---|---|
| Markdown (text) cell | formatted text, headings, links, equations, tables and images |
| Code cell | executable Python code |

In Colab you run a code cell with the **play button** on its left, or with **Shift + Enter**.

A notebook is connected to a Python process called the **runtime** (or **kernel**). The runtime keeps your variables in memory while it is active. If the runtime restarts, the variables are gone and you must run the relevant cells again.

### Things to know about Colab

- **You need a Google account.** UCL runs on Microsoft 365, not Google Workspace, so most of you will use a personal Google account. That is fine. Later in the term the same account will be used for **Google Earth Engine**.
- **The runtime is temporary.** Colab disconnects after a period of inactivity, and there is also a maximum session length. When that happens, **all variables and all files in `/content` are deleted**.
- **Your notebook is saved to Google Drive, but your data files are not.** Anything you download or create during a session lives on a temporary virtual machine unless you save it to Drive (see 2.3).

### Good notebook habits

- Run notebooks **from top to bottom**.
- Keep code cells short, with one clear purpose each.
- Use text cells to explain **why** you are doing something.
- Use clear variable names (`ndvi_2024`, not `x2`).
- Do not rely on hidden state from cells you ran earlier or out of order.
- Before sharing or submitting: **Runtime → Restart session and run all**. This is also the first item on the Part B submission checklist.

Useful menu options: **File → Save a copy in Drive**, **File → Download → Download .ipynb**, **File → Print** (to PDF), **Runtime → Run all**, **Runtime → Restart session and run all**.

### Example: lists and methods

A **list** stores multiple values. A **method** is a function attached to an object, called with a dot: `my_list.append(...)`.

Below, the list holds Sentinel-2 band names:

- `append()` adds **one object** to the end of a list;
- `clear()` removes all items from a list.

````python
# Sentinel-2 visible bands: blue, green, red
bands = ["B02", "B03", "B04"]
# Near-infrared and SWIR bands we might want to add
extra_bands = ["B08", "B11"]

# append() adds extra_bands as ONE object (a list inside a list)
bands.append(extra_bands)
print("bands =", bands)
print("number of items:", len(bands))

# clear() empties extra_bands
extra_bands.clear()
print("extra_bands =", extra_bands)
````

Notice that `bands` now has **4** items, not 5, and the last item is itself a list. This is often not what you want, which leads to the next exercise.

### `!cd` vs `%cd`

Each `!` command runs in its **own, short-lived shell**. So `!cd some_folder` changes directory and then that shell closes; the next cell is back where it started. To change the directory of the **notebook itself**, use the magic command `%cd`.

This is the same idea as in section 2.2: a runtime keeps state, and a throw-away process does not.

````python
!cd sample_data 2>/dev/null; pwd   # the directory changes only inside this one shell
!pwd                               # ...and here we are back where we started
````
