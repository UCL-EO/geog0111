## 2.5. Earth Observation data – Sentinel-2 (NDVI / NDBI) and Landsat (LST)

Typical choices to calculate NDVI, NDBI or actual temperature are:

- **Sentinel-2 Surface Reflectance (S2_SR)** for NDVI, NDBI and land cover patterns.  
- **Landsat 8/9 Level-2** for Land Surface Temperature (LST).

Document all important choices (bands, dates, cloud threshold, etc.).

**Useful details (check them on the dataset pages):**

- Sentinel-2 SR: multiply the bands by 0.0001 to get reflectance. NDVI = (B8 − B4) / (B8 + B4); NDBI = (B11 − B8) / (B11 + B8). B11 is 20 m; B8 and B4 are 10 m.
- Mask clouds before you build a composite (e.g. Cloud Score+, as in Week 6).
- Landsat Collection 2 Level-2 LST: `ST_B10 × 0.00341802 + 149.0` gives kelvin; subtract 273.15 for °C. Mask clouds and cloud shadows with `QA_PIXEL` (bits 1, 3 and 4).
- Compare the **same season** in both years (e.g. June–August), and check whether either year was unusually hot or dry (e.g. summer 2022). One extreme summer can look like a trend.
- LST is the temperature of the **surface**, not of the air. Explain how this affects any comparison with the station data.
