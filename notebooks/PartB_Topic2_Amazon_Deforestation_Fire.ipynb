**Forest loss – Hansen Global Forest Change**

https://developers.google.com/earth-engine/datasets/catalog/UMD_hansen_global_forest_change_2025_v1_13

- **Key bands:**  
  - `treecover2000` – percent tree cover in 2000  
  - `lossyear` – year of forest loss (coded 1 = 2001 … 25 = 2025; pixels with no loss are masked)  
- **Purpose in this topic:**  
  - Define **forest in 2000** using a tree cover threshold  
  - Compute **deforested area (km²)** in the Amazon Basin for your two chosen years  
  - Produce **deforestation maps**
- **Tip:** summing 30 m pixels over the whole basin is a very large computation and can time out. Use `ee.Image.pixelArea()`, keep `scale=30`, and add `maxPixels=1e13` and `tileScale` to `reduceRegion`, or run it as `Export.table.toDrive`. Setting a coarser `scale` makes it faster but changes the numbers, because Earth Engine then works on resampled versions of the bands.

**Burned area – MODIS MCD64A1**

https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MCD64A1

  - The product is **monthly** at 500 m. For an annual burned mask, filter the year, turn each month into burned / not burned (`BurnDate > 0`) and take the `max()`.  
- **Purpose in this topic:**  
  - Compute **burned area (km²)** in the Amazon Basin for each of your two chosen years  
  - Produce **burned-area maps** and compare spatial patterns
