## 1.2 `if ... elif ... else`

A conditional statement runs a block of code **only if** a condition is `True`:

```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if none of the above were True
```

- `elif` and `else` are optional. You can have as many `elif`s as you need.
- Python checks the conditions **in order** and runs **only the first** block whose condition is True.
- The line before each block ends with a **colon** `:`.
- The block is defined by its **indentation**. Use **4 spaces** (Colab does this automatically when you press Enter after a `:`). Getting indentation wrong is the most common beginner error in Python.

An EO example: classifying a pixel by its NDVI value. The thresholds below are **illustrative**; real thresholds depend on the sensor, the season and the region.

````python
ndvi = 0.63

if ndvi < 0:
    land_class = 'water'
elif ndvi < 0.2:
    land_class = 'bare soil / built-up'
elif ndvi < 0.5:
    land_class = 'sparse vegetation'
else:
    land_class = 'dense vegetation'

print(f'NDVI {ndvi} -> {land_class}')
````

Why does `elif ndvi < 0.2` not need `ndvi >= 0`? Because we only get there if the first condition was **False**, so we already know `ndvi >= 0`. The **order** of the conditions matters.

## 2.1 `for ... in ...`

A `for` loop runs a block of code **once for each item** in a group (a list, tuple, string, dictionary, `range`, ...). The loop variable takes each value in turn.

````python
countdown = [4, 3, 2, 1]

for item in countdown:
    print(item)

print('lift-off!')       # NOT indented: runs once, after the loop
````

Compare it with the version below, where the last line **is** indented. It is now **inside** the loop and runs every time:

````python
countdown = [4, 3, 2, 1]

for item in countdown:
    print(item)
    print('lift-off!')   # indented: runs on every pass of the loop
````

## 2.2 Accumulating results

A very common pattern is to start with an empty or zero value **before** the loop, and update it **inside** the loop:

````python
rain_mm = [55.2, 40.1, 38.6, 43.0, 49.3, 45.1, 44.8, 50.2, 49.5, 68.7, 59.0, 55.6]

total = 0                 # initialise before the loop
wet_months = []           # an empty list to collect results

for r in rain_mm:
    total += r            # running total
    if r > 50:
        wet_months.append(r)

print(f'total = {total:.1f} mm (check with sum(): {sum(rain_mm):.1f})')
print('months above 50 mm:', wet_months)
````

## 2.3 Counting: `enumerate()` and `range()`

Often you need the **position** of each item as well as its value. You could keep your own counter, but `enumerate()` does it for you:

````python
months = ['Jan', 'Feb', 'Mar']

# manual counter
count = 0
for m in months:
    print(f'count {count}: {m}')
    count += 1

# the same with enumerate()
for i, m in enumerate(months):
    print(f'index {i}: {m}')
````

`range(start, stop, step)` produces a sequence of integers: from `start` (default 0) **up to but not including** `stop`, in steps of `step` (default 1).

````python
print(list(range(6)))          # 0..5
print(list(range(2, 10, 2)))   # 2, 4, 6, 8 (10 is NOT included)

for year in range(2020, 2025):
    print(year, end=' ')       # end=' ' prints a space instead of a new line
````

## 2.4 Looping over two lists and over dictionaries

`zip()` walks through two (or more) lists **side by side**. For a dictionary, `.items()` gives `(key, value)` pairs.

`assert condition` stops the code with an `AssertionError` if the condition is False. It is a quick way to **check your assumptions** before going on.

````python
bands = ['B02', 'B03', 'B04', 'B08']
wavelengths = [490, 560, 665, 842]

assert len(bands) == len(wavelengths), 'lists must have the same length'

for b, wl in zip(bands, wavelengths):
    print(f'{b}: {wl} nm')

s2_wavelength = dict(zip(bands, wavelengths))
for band, wl in s2_wavelength.items():
    if wl > 700:
        print(f'{band} is in the near-infrared ({wl} nm)')
````

## 2.5 List comprehensions

Building a list with a loop is so common that Python has a compact syntax for it:

```python
new_list = [expression for item in old_list]
new_list = [expression for item in old_list if condition]
```

````python
# the long way
squares = []
for x in range(5):
    squares.append(x * x)
print(squares)

# the same as a list comprehension
print([x * x for x in range(5)])

# with a condition: keep only clear scenes
cloud = [80.5, 12.0, 45.3, 3.2, 19.9, 66.0]
clear = [c for c in cloud if c < 20]
print('clear scenes:', clear)
````

Use comprehensions for **simple** expressions. If you need several lines of logic, a normal loop is easier to read, and *readability counts*. Run `import this` for Python's philosophy in 19 lines.

## 2.6 `while`, `break` and `continue`

A `while` loop repeats **as long as** a condition is True. It is useful when you don't know in advance how many repeats you need. **Make sure the condition eventually becomes False**, or the loop runs forever (in Colab, stop it with the ■ button or *Runtime → Interrupt execution*).

- `break` leaves the loop immediately.
- `continue` skips the rest of the current pass and moves on to the next one.

````python
# daily cloud cover over a site: find the first clear day
cloud = [80.5, 65.0, 92.1, 40.2, 12.0, 70.3, 5.5]

day = 0
while cloud[day] >= 20:
    day += 1
print('first clear day (while):', day, '->', cloud[day], '%')

# the same with for + break
for day, c in enumerate(cloud):
    if c < 20:
        print('first clear day (break):', day)
        break

# continue: skip cloudy days, process only the clear ones
for day, c in enumerate(cloud):
    if c >= 20:
        continue
    print('processing day', day)
````

````python
def hello_world():
    """Print the string 'hello world'."""
    print('hello world')
````

## 3.2 Inputs and outputs: arguments and `return`

Most useful functions take **arguments** and **return** a result. Think of a function as a **filter**: inputs go in, an output comes out.

```python
def function_name(arg1, arg2):
    """One-line summary of what the function does.

    Parameters
    ----------
    arg1 : type
        What it is (with units!).
    arg2 : type
        What it is.

    Returns
    -------
    type
        What comes back.
    """
    result = ...      # do the work
    return result
```

The docstring layout above (the "NumPy style") is the one used by `numpy`, `pandas` and most scientific packages, and it is a good model for Part B.

````python
def compute_ndvi(nir, red):
    """Calculate the Normalised Difference Vegetation Index.

    Parameters
    ----------
    nir : float
        Near-infrared surface reflectance (0-1).
    red : float
        Red surface reflectance (0-1).

    Returns
    -------
    float
        NDVI, in the range -1 to 1.
    """
    return (nir - red) / (nir + red)


park = compute_ndvi(0.42, 0.05)
car_park = compute_ndvi(nir=0.22, red=0.18)   # arguments can also be given by name
print(f'park: {park:.2f}, car park: {car_park:.2f}')
````

### `return` versus `print`

A very common mistake is to `print` the result inside a function instead of `return`ing it. A printed value is just shown on the screen. A returned value can be **stored and used** in the rest of your workflow.

````python
def ndvi_printing(nir, red):
    """Print NDVI (a poor design: nothing is returned)."""
    print((nir - red) / (nir + red))

value = ndvi_printing(0.42, 0.05)
print('stored value:', value)          # None: the function returned nothing
````

> **Part B rule:** every function must have clear input arguments, **return** a value (not just print it), and include a short docstring.

## 3.3 Testing functions with `assert`

## 3.4 Designing a function

A reliable way to write any function:

1. **Purpose:** state in one sentence what it should do.
2. **Inputs and output:** what goes in and what comes out? Write down an example input and the output you expect.
3. **Skeleton:** write the `def` line and the docstring, and simply `return` the input unchanged. Check that it runs.
4. **Core code:** try the key step on an example **outside** the function first, then put it inside.
5. **Test:** add `assert` statements.
6. **Improve:** what inputs could break it? Add checks.

Example: a function to combine a list of name parts into one string.

````python
# Step 3: skeleton
def full_name(name_list):
    """Join a list of name parts into a single string.

    Parameters
    ----------
    name_list : list of str
        Parts of the name, e.g. ['Ada', 'Smith'].

    Returns
    -------
    str
        The parts joined by single spaces.
    """
    retval = name_list          # placeholder: pass the input straight through
    return retval

print(full_name(['Ada', 'Smith']))
````

````python
# Step 4: try the core idea outside the function
print(' '.join(['Ada', 'Smith']))

# Step 4-5: put it in the function and test
def full_name(name_list):
    """Join a list of name parts into a single string.

    Parameters
    ----------
    name_list : list of str
        Parts of the name, e.g. ['Ada', 'Smith'].

    Returns
    -------
    str
        The parts joined by single spaces.
    """
    return ' '.join(name_list)

assert full_name(['Ada', 'Smith']) == 'Ada Smith'
assert full_name(['Ada', 'King', 'Smith']) == 'Ada King Smith'
print('tests passed')
````

## 3.5 Default and keyword arguments

Arguments can have **default values**. The caller can leave them out, or override them by **name**. This is how most library functions work, for example `print(..., sep=' ', end='\n')`:

````python
print('B02', 'B03', 'B04')
print('B02', 'B03', 'B04', sep=', ')
print('B02', 'B03', 'B04', sep='_', end=' <- joined\n')
````

A real EO example: Sentinel-2 Level-2A pixels are stored as integers (DN). To get surface reflectance:

$$\rho = \frac{DN + \text{offset}}{\text{quantification}}$$

with a quantification value of 10000. Products processed with **baseline 04.00 or later** (from January 2022) also have an offset of −1000; older products have no offset. A keyword argument with a sensible default handles both cases:

````python
def dn_to_reflectance(dn, quantification=10000, offset=-1000):
    """Convert a Sentinel-2 L2A digital number to surface reflectance.

    Parameters
    ----------
    dn : int or float
        Digital number stored in the L2A product.
    quantification : int, optional
        Quantification value (default 10000).
    offset : int, optional
        Additive offset (default -1000, for processing baseline >= 04.00;
        use 0 for older products).

    Returns
    -------
    float
        Surface reflectance (usually 0-1).
    """
    return (dn + offset) / quantification


print(dn_to_reflectance(5200))              # new products: (5200 - 1000) / 10000
print(dn_to_reflectance(5200, offset=0))    # older products
````

Rules: in the `def` line, arguments with defaults come **after** those without. When calling, positional arguments come **first**, then keyword arguments, in any order.

## 3.6 Scope: functions should not depend on notebook variables

Variables created **inside** a function are **local**: they exist only while the function runs. A function *can* read variables from the notebook (**global** variables), but relying on that is risky. The function's result then depends on whatever that variable happens to hold at the time, which in a notebook depends on which cells you ran, and in which order.

````python
scale = 10000

def to_reflectance_bad(dn):
    """Convert DN to reflectance (BAD: depends on a global variable)."""
    return dn / scale                    # where does 'scale' come from?

print(to_reflectance_bad(5000))

scale = 100                              # ... much later, someone reuses the name
print(to_reflectance_bad(5000))          # same input, different answer!

def to_reflectance_good(dn, scale=10000):
    """Convert DN to reflectance; everything it needs is an argument."""
    return dn / scale

print(to_reflectance_good(5000))         # always the same answer
````

````python
def demo():
    """Show that variables created inside a function are local."""
    inside = 'I only exist inside demo()'
    return inside

demo()
print('inside' in globals())             # False: the local variable has gone
````

> **Part B rule:** functions must *"have clear input arguments (not rely on notebook variables)"*. If a function needs a value, pass it in as an argument.

## 3.8 `*args` and `**kwargs` (to recognise)

In documentation you will often see `function(*args, **kwargs)`:

- `*args` collects any number of **positional** arguments into a tuple;
- `**kwargs` collects any number of **keyword** arguments into a dictionary.

The same `*` and `**` can **unpack** a list or dictionary when calling a function. You won't need to write these yourself yet, but you should recognise them.

````python
bands = ['B02', 'B03', 'B04']
print(*bands)                         # same as print('B02', 'B03', 'B04')

options = {'sep': ' | ', 'end': ' <- end\n'}
print(*bands, **options)              # same as print(..., sep=' | ', end=' <- end\n')
````

## 3.9 `lambda`, `map()` and `reduce()`

These three ideas are central to **Google Earth Engine** (Week 8), where you apply a function to every image in a collection (`collection.map(...)`) and combine many images into one (`collection.reduce(...)`).

- A **lambda** is a small, nameless one-line function: `lambda x: 2 * x`.
- `map(function, iterable)` applies a function to every item.
- `reduce(function, iterable)` combines the items **pairwise** into a single result.

````python
t_k = [271.2, 275.8, 283.4, 290.1, 295.6]

# map with a named function...
print(list(map(lst_to_celsius, t_k)))

# ...or with a lambda
print(list(map(lambda t: round(t - 273.15, 1), t_k)))

# the list comprehension equivalent (usually the clearest in plain Python)
print([round(t - 273.15, 1) for t in t_k])
````

````python
from functools import reduce

def add_and_report(a, b):
    print(f'combining {a} and {b}')
    return a + b

print('total:', reduce(add_and_report, [1, 3, 7]))
print('maximum:', reduce(lambda a, b: a if a > b else b, t_k))
````

## 3.10 Style: PEP 8 and PEP 257

Python has official style guides. Following them makes your code easier for others (and your markers) to read:

- [PEP 8](https://peps.python.org/pep-0008/): code style. Use 4-space indents, `snake_case` names, spaces around `=` in assignments (`x = 1`), but **not** around `=` in keyword arguments (`f(x, offset=0)`), and lines of up to 79 characters.

## 4.1 Writing a `.py` file from Colab

The cell magic `%%writefile filename` saves the rest of the cell to a file instead of running it.

````python
%%writefile eo_utils.py
"""Small utility functions for Earth Observation exercises (GEOG0111)."""


def compute_ndvi(nir, red):
    """Calculate NDVI from near-infrared and red reflectance.

    Returns None if nir + red == 0.
    """
    if nir + red == 0:
        return None
    return (nir - red) / (nir + red)


def kelvin_to_celsius(t_kelvin, offset=273.15):
    """Convert a temperature from kelvin to degrees Celsius."""
    return t_kelvin - offset


if __name__ == '__main__':
    # This block runs only when the file is executed as a script,
    # not when it is imported. Use it for quick self-tests.
    assert compute_ndvi(0.5, 0.5) == 0
    assert kelvin_to_celsius(273.15) == 0
    print('eo_utils self-tests passed')
````

# Summary

| Topic | Key points |
|---|---|
| `if` / `elif` / `else` | condition, colon, **4-space indent**; only the first True branch runs; order matters |
| `for` | `for item in group:`; `range(start, stop, step)` (stop excluded); `enumerate()`; `zip()`; `dict.items()` |
| Accumulating | initialise before the loop, update inside it |
| Comprehensions | `[f(x) for x in xs if cond]`: short and readable, for simple cases |
| `while` | repeats while the condition is True; make sure it ends; `break`, `continue` |
| `assert` | `assert condition, 'message'`: check assumptions and test functions |
| Functions | `def name(args):` + docstring + `return`; test with `assert` |
| Arguments | defaults `f(x, offset=0)`; positional first, then keyword; `*args`, `**kwargs` |
| Scope | variables inside a function are local; **pass everything in as arguments** |
| `lambda`, `map`, `reduce` | small functions and applying them to collections, as in Earth Engine |
| Modules | `%%writefile x.py`, `import x`, `importlib.reload(x)`, `!python x.py`, `if __name__ == '__main__':` |

**Part B checklist for every function:** clear arguments (no notebook variables), a `return` value, a docstring, and at least one test.
